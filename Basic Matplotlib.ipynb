# Load Necessary Libraries

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Basic Graph

```python
x = [0,1,2,3,4]
y = [0,2,4,6,8]

# Resize your Graph (dpi specifies pixels per inch. When saving probably should use 300 if possible)
plt.figure(figsize=(5,3),dpi=100)

# Line 1

# Keyword Argument Notation
# plt.plot(x,y, label='2x',color='red',linewidth=2,marker='.',markersize = 10,markeredgecolor = 'blue',linestyle='--')

# Shorthand notation
# fmt = '[color][marker][line]'
plt.plot(x,y,'r^--', label='2x')


# Line 2

# Select intervals we want to plot points at
x2 = np.arange(0,4.5,0.5)

# plot part of the graph as line

# plt.plot(x2,x2**2,'b',label='X^2')           # Basic line 2

plt.plot(x2[:6],x2[:6]**2,'b',label='X^2')     # Decorated line 2 (by using array slicing)
# plot remaining of the graph as dotted line
plt.plot(x2[5:],x2[5:]**2,'b--')


# Add a title (specify font parameters with fontdict)
# plt.title('Our First Graph!', fontdict={'fontname': 'Comic Sans MS', 'fontsize': 20})
plt.title('My First Graph!',fontdict={'fontsize':20})

# X, Y axis Tickmarks (scale of your graph)
plt.xticks([0,1,2,3,4])
# plt.yticks([0,2,4,6,8,10])

# X and Y Labels
plt.xlabel('X Axis')
plt.ylabel('Y Axis')

# Save figure (dpi 300 is good when saving so graph has high resolution)
plt.savefig("MyGraph.png",dpi=300)

# Add a legend
plt.legend()

# Show plot
plt.show()
```

# Bar Chart

```python
labels = ['A','B','C']                    # X axis
values = [1,4,2]                          # Y axis

plt.title("My Bar Chart")

bars = plt.bar(labels,values)               # Nothing but `plt.bar(x,y)`

plt.figure(figsize=(6,4))

# bars[0].set_hatch('/')
# bars[1].set_hatch('o')
# bars[2].set_hatch('*')

# We can also use 'for loop' for the above
patterns = ['/','o','*']
for bar in bars:
  bar.set_hatch(patterns.pop(0))

# Saving the pic
plt.savefig('MyBarChart.png',dpi=300)

# Show fig
plt.show()
```

# Histograms

```python
bins = [40,50,60,70,80,90,100]

plt.figure(figsize=(8,5))

plt.hist(fifa.Overall,bins=bins,color='#abefef')

plt.xlabel('Number of Players')
plt.ylabel('Skill level')
plt.title('Distribution of Player Skills in FIFA 2018')

plt.xticks(bins)

plt.savefig('Histogram.png',dpi=300)

plt.show()
```

# Pie Chart

```python
left = fifa.loc[fifa['Preferred Foot'] == 'Left'].count().iloc[0].item()
right= fifa.loc[fifa['Preferred Foot'] == 'Right'].count().iloc[0].item()

plt.figure(figsize=(8,5))

labels = ['Left','Right']
colors = ['#873e23','#e28743']

plt.pie([left,right],labels=labels,colors=colors, autopct='%.2f %%')

plt.title('Foot Preferrence by FIFA Players ')

plt.savefig('PieChart.png',dpi=300)

plt.show()
```

# Pie Chart #2

```python
fifa.Weight = [int(x.strip('lbs')) if (type(x) == str) else x for x in fifa.Weight]

plt.figure(figsize=(8,5),dpi=100)

plt.style.use('ggplot')

light = fifa.loc[fifa.Weight < 125].count().iloc[0].item()
light_medium = fifa.loc[(fifa.Weight >= 125) & (fifa.Weight < 150)].count().iloc[0].item()
medium = fifa.loc[(fifa.Weight >=150) & (fifa.Weight < 175)].count().iloc[0].item()
medium_heavy = fifa.loc[(fifa.Weight >=175) & (fifa.Weight < 200)].count().iloc[0].item()
heavy = fifa.loc[fifa.Weight >= 200].count().iloc[0].item()

Weights = [light,light_medium,medium,medium_heavy,heavy]
labels = ['Less than 125','125-150','150-175','175-200','More than 200']
explode = [.4,.1,0,0,.4]

plt.pie(Weights,labels=labels,explode=explode, autopct='%.2f %%')

plt.title('Weight of the professional Footballers (in lbs)')


plt.show()
```

# Scatter Plots

```python
# Dribling vs Finishing
plt.title('Dribbling vs Finishing')

x = fifa.Dribbling[(fifa.Dribbling > 70) & (fifa.Dribbling < 100)].head(50)

# x[['Name','Dribbling']]

y = fifa.Finishing[(fifa.Finishing > 70) & (fifa.Finishing < 100)].head(50)

# y[['Name','Finishing']]

plt.xlabel('Driblling')
plt.ylabel('Finishing')

plt.scatter(x,y, color='blue',edgecolors='k',linewidths=1)

# x = fifa.loc[fifa['Name'] == 'L. Messi']
# x[['Name','Dribbling','Finishing']]          # Simply the best player of all time :)

plt.show()
```

# Box and Whiskers charts


```python
barca = fifa.loc[fifa.Club == 'FC Barcelona']['Overall']
madrid = fifa.loc[fifa.Club == 'Real Madrid']['Overall']
Man_City = fifa.loc[fifa.Club == 'Manchester City']['Overall']
Al_Hilal = fifa.loc[fifa.Club == 'Al Hilal']['Overall']

labels = ['FC Barcelona','Real Madrid','Manchester City','Al Hilal']

bp = plt.boxplot([barca,madrid,Man_City,Al_Hilal],tick_labels=labels,patch_artist=True,  medianprops={'linewidth': 2})

for box in bp['boxes']:
  # change outline color
  box.set(color='#154c79',linewidth=2)
  # change fill color
  box.set(facecolor='#abdbe3')

plt.title('Professional Football teams Comparison')
plt.ylabel('FIFA Overall Rating')


plt.show()
```
